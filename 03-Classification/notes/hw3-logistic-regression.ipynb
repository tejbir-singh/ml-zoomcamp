```python
categorical = list(df.dtypes[df.dtypes == 'str'].index)
categorical
```

```python
for c in categorical:
    df[c] = df[c].str.lower().str.replace(' ', '_')
    df[c] = df[c].fillna('NA')
    print(df[c].isna().sum())
```

```python
df.dtypes
```

```python
numerical = ['annual_income', 'number_of_courses_viewed', 'interaction_count', 'lead_score']
for col in numerical:
    print(df[col].isna().sum())
```

```python
df['lead_score'].isna().sum()
```

```python
df['annual_income'] = df['annual_income'].fillna(0)
```

```python
df['lead_score'] = df['lead_score'].fillna(0)
```

```python
from sklearn.model_selection import train_test_split
# split the data
df_full_train, df_test = train_test_split(df, test_size=0.2, random_state=42)
df_train, df_val = train_test_split(
    df_full_train, test_size=0.25, random_state=42
)
```

```python
df_train = df_train.reset_index(drop=True)
df_val = df_val.reset_index(drop=True)
df_test = df_test.reset_index(drop=True)
```

```python
# extract the target variable or `converted` column
y_train = df_train.converted.values
y_val = df_val.converted.values
y_test = df_test.converted.values
```

```python
# EDA
# Use the training and validation dataset together
df_full_train = df_full_train.reset_index(drop=True)
```

```python
# One-hot encode the categoriacl columns into multiple binary columns
dicts = df[categorical + numerical].to_dict(orient='records')
```

```python
from sklearn.feature_extraction import DictVectorizer
```

```python
dv = DictVectorizer(sparse=False)
all_categories = categorical + numerical
all_categories

train_dict = df_train[categorical + numerical].to_dict(orient='records')
X_train = dv.fit_transform(train_dict)
```

```python
# get the names of all the columns
dv.get_feature_names_out()
```

```python
X_train.shape
```

```python
val_dict = df_val[categorical + numerical].to_dict(orient='records')
X_val = dv.transform(val_dict)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(solver='liblinear', C=1.0, max_iter=1000, random_state=42)
model.fit(X_train, y_train)
```

```python
model.intercept_[0]
```

```python
model.coef_[0].round(3)
```

```python
y_pred = model.predict_proba(X_val)[:, 1]
```

```python
converted_decision = (y_pred >= 0.5)
```

```python
# measuring accuracy of the trained model on the validation set
(y_val == converted_decision).mean()
```
