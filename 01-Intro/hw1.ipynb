```python
# Returns a Series with the count of missing values for each column
df.isna().sum()
```

```python
# Returns MAX 'fuel_efficiency_mpg' for all cars manufactured in Asia
max_fuel_efficiency_value_asia = df.loc[df['origin'] == 'Asia', 'fuel_efficiency_mpg'].max()
print(max_fuel_efficiency_value_asia)
```

```python
# Q6. Median value of horsepower
median_value_hp = df['horsepower'].median()
print(median_value_hp)
```

```python
# find the most frequent occurring value in the 'horsepower' column
most_frequent_hp = df['horsepower'].mode()[0]
print(most_frequent_hp)
```

```python
# Fill missing values in the 'horsepower' column with the most frequent occuring value
df['horsepower'] = df['horsepower'].fillna(most_frequent_hp)
```

```python
# recalculate the median value of the 'horsepower' col
median_value_hp = df['horsepower'].median()
print(median_value_hp)
```

```python
# Q7 Sum of weights
# select all cars from Asia
df_filtered_cars_asia = df.loc[df['origin'] == 'Asia']
```

```python
# Select only cols 'vehicle_weight' & 'model_year'
df_subset_cols_vh_weight_model_year = df_filtered_cars_asia[['vehicle_weight', 'model_year']]
print(df_subset_cols_vh_weight_model_year)
```

```python
X = df_subset_cols_vh_weight_model_year.head(7).to_numpy()
print(X)
```

```python
# Multiply the matrix X with its transpose
XTX = X.T.dot(X)
print(XTX)
```

```python
# Invert the df XTX
XTX_inverted = np.linalg.pinv(XTX)
print(XTX_inverted)
```

```python
# Create a pandas ExtensionArray
y = pd.array([1100, 1300, 800, 900, 1000, 1100, 1200])
print(y)
```
